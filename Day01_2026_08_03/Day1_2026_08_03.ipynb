```python
cname = input("Enter Customer name: ")

amount = float(input("Enter purchase amount: "))

print("Customer name:",cname)
print("purchase amount:",amount)
if amount >= 1000:
    discount = amount / 10
    print("Discount:",discount)
    final = amount - discount
    print("Final amount:",final)
else:
    print("No Discount")
    
```
