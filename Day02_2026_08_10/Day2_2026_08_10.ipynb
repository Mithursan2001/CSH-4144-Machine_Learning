```python
#Functions
def calculate_average1(a,b):
    return (a+b)/2
def calculate2(a,b):
    return (a-b)
def calculate3(a,b):
    return (a*b)
def calculate4(a,b):
    return (a/b)
def calculate5(a,b):
    return (a+b)

result1 = calculate_average1(20,30)
print(result1)
result2 = calculate2(40,30)
print(result2)
result3 = calculate3(20,30)
print(result3)
result4 = calculate4(60,30)
print(result4)
result5 = calculate5(60,30)
print(result5)
```
